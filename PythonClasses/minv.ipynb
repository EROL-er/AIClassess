```python
my_list = [1, 2, 3, 4, 5, 2, 3, 6]
seen = []
duplicates =[]

for i in my_list:
    if i in seen:
        duplicates.append(i)
    else:
        seen.append(i)

print(f"Duplicate elements: {list(duplicates)}")
```

```python
from collections import Counter

my_list = [1, 2, 3, 4, 5, 2, 3, 6]
counts = Counter(my_list)
duplicates = {item: count for item, count in counts.items() if count > 1}
print(f"Duplicate elements and their counts: {duplicates}")


flot=121.999999
inr=int(flot)
print(inr)
```

```python

from collections import Counter
l = [1, 2, 3, 4, 4, 2]
dupes = [item for item, count in Counter(l).items() if count > 1]
print(dupes)   # [2, 4]
```
