```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder, PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, classification_report,confusion_matrix, ConfusionMatrixDisplay
from sklearn.linear_model import ElasticNet,Lasso,Ridge
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import mutual_info_regression, SelectKBest
from scipy import stats
from sklearn.feature_selection import f_regression
from sklearn.datasets import load_breast_cancer
from collections import Counter
from sklearn.datasets import load_iris, load_digits
from sklearn.svm  import SVC
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

%matplotlib inline
iris = load_iris()
```

```python

iris.feature_names

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target']= iris.target
iris.target_names
df.tail()
df['flower_name'] = df.target.apply(lambda x: iris.target_names[x])


X = df.drop(['target', 'flower_name'], axis='columns')
y = df['target']

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.20)
model = SVC(kernel='linear')

model.fit(x_train, y_train)
model.score(x_test, y_test)

sample_df = pd.DataFrame([[6.8, 3.0, 5.2, 2.2]], columns=iris.feature_names)
pred = model.predict(sample_df)[0]
print(iris.target_names[pred])

```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)

df_1500 = df.loc[df.index.repeat(10)].reset_index(drop=True)

df_1500["target"] = np.repeat(iris.target, 10) 

df_1500['flower_name'] = df_1500.target.apply(lambda x: iris.target_names[x])

df_shuffled = df_1500.sample(frac=1, random_state=42).reset_index(drop=True)

X = df_shuffled.drop(["target", 'flower_name'], axis='columns')
y = df_shuffled["target"]

X_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.30)

dt = SVC()

dt.fit(X_train, y_train)

y_pred = dt.predict(x_test)

score=classification_report(y_test, y_pred)


print(score)

df_1500
```
