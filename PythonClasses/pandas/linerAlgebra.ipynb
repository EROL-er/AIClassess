```python
# 🧠 Each user represented by [age, monthly_purchases, time_spent_on_app(mins)]
users = np.array([
    [25, 3, 40],
    [27, 5, 42],
    [35, 10, 15],
    [23, 4, 41],
    [60, 1, 10],
    [26, 3, 38],
])

user_names = np.array(["Alice", "Bob", "Charlie", "Diana", "Edward", "Frank"])

# 🧍 New user vector
new_user = np.array([26, 4, 39])

# 🧮 Compute distance using np.linalg.norm for each user
distances = np.array([np.linalg.norm(u - new_user) for u in users])

# Combine into DataFrame
df = pd.DataFrame({
    'User': user_names,
    'Distance': distances
}).sort_values(by='Distance')

# 🎯 Show nearest 3 neighbors
print(df.round().head(3))
```

```python

# Define a vector
v = np.array([250, 40, 5,5,8,900,8,300,7])
print(f"Original vector: {v}")

# Compute the L2 norm (vector length)
v_norm = np.linalg.norm(v)
print(f"L2 Norm: {v_norm}")

# Normalize the vector (create a unit vector)
v_unit = v / v_norm
print(f"Unit vector: {v_unit}")

# Round and print values for readability
for i in v_unit:
    print(round(i))
```

```python

users = np.array([
    [25, 3, 40],
    [27, 5, 42],
    [35, 10, 15],
    [23, 4, 41],
    [60, 1, 10],
    [26, 3, 38],
])

user_names = np.array(["Alice", "Bob", "Charlie", "Diana", "Edward", "Frank"])

# 🧍 New user vector
new_user = np.array([26, 4, 39])


distances_raw = np.array([np.linalg.norm(u - new_user) for u in users])

df_raw = pd.DataFrame({
    'User': user_names,
    'Distance (Raw)': distances_raw
}).sort_values(by='Distance (Raw)')


users_min = users.min(axis=0)
users_max = users.max(axis=0)

users_norm = (users - users_min) / (users_max - users_min)
new_user_norm = (new_user - users_min) / (users_max - users_min)

# Compute distances again
distances_norm = np.array([np.linalg.norm(u - new_user_norm) for u in users_norm])

df_norm = pd.DataFrame({
    'User': user_names,
    'Distance (Normalized)': distances_norm
}).sort_values(by='Distance (Normalized)')


df_compare = pd.concat([df_raw.set_index('User'), df_norm.set_index('User')], axis=1)
print(df_compare.sort_values(by='Distance (Normalized)'))
```

```python
# Identifier is a stablizer. 

A = np.array([[2, 4],
              [4, 8]])  # determinant = 0 (singular)

I = np.eye(2)
A_fixed = A + 0.1 * I   # stabilize it

print(np.linalg.det(A))       # 0.0 ❌
print(np.linalg.det(A_fixed)) # 0.4 ✅
print(A_fixed)
```

```python
X = np.array([[1, 1],
              [1, 2],
              [1, 3]])
y = np.array([2, 2.5, 3.5])

w = np.linalg.inv(X.T @ X) @ (X.T @ y)
print(w)
```
