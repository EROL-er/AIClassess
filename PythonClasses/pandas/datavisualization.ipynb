```python
# Pyplot API
# Univariable - Numerical

data = {'Salary':[25000, 30000, 37000, 28000,39000,48000, 55000, 52000, 35000,26000, 27000, 31000, 85000, 90000, 9500, 1000,8500, 9000, 95000, 1000]}

df = pd.DataFrame(data)
df.head(5)
df.shape
```

```python
# Line Plot
df["experience"] = [1, 2, 4, 1.5, 5, 8, 12, 10, 4, 1, 1.2, 2, 3, 1.8, 5.5, 4.5, 3.2, 7, 8, 10]
df["Age"] = [22, 25,29, 24,30, 35,40,36,28, 22, 23, 25, 26, 24, 31, 29, 27, 33, 35, 38]
df['dept'] = ['HR', 'IT', 'Finance', 'HR', 'Finance', 'IT', 'HR', 'Finance','IT', 'HR'] * 2
plt.plot(df['Salary'], color='red', marker='o', linestyle= ":", linewidth= "2")
plt.show
```

```python
plt.hist(df['Salary'], bins=5, color='green')
plt.show
```

```python



#Pie Char
acount = df['dept'].value_counts()

plt.pie(acount,labels=acount.index,autopct='%1.1f',explode=[0,0.1,0.2])
plt.axis('equal')
plt.show
```

```python
plt.bar(acount.index,acount, color=['salmon', 'skyblue', 'green'])
plt.show
```

```python
# Bivariate

plt.scatter(df['Age'],df['Salary'],color='orange')
plt.show
```

```python
# Line plot
sorted_age= df.sort_values('Age')
plt.plot(sorted_age['Age'],df['Salary'], color='red', marker='*', linestyle= ':', linewidth= '2')
plt.show
```

```python
plt.bar(sorted_age['Age'],df['Salary'],color='green', align='center')
plt.show
```

```python
hr_sal = df[df['dept']== 'HR']['Salary']
it_sal = df[df['dept']== 'IT']['Salary']
fin_sal = df[df['dept']== 'Finance']['Salary']
plt.boxplot([hr_sal, it_sal, fin_sal], labels=['HR', 'IT', 'Finance'])
plt.legend()
plt.show
```

```python
salary_by_dept = df.groupby('dept')['Salary'].sum()
plt.pie(salary_by_dept, labels=salary_by_dept.index, autopct='%1.2f',explode=[0,0.1,0.1], shadow=True)
plt.axis()
plt.show
```

```python
hr_mean = sum(hr_sal)/len(hr_sal)
it_mean = sum(it_sal)/len(it_sal)
fin_mean = sum(fin_sal)/len(fin_sal)
plt.bar(["HR", "IT", "Finance"], [hr_mean, it_mean, fin_mean], color = ["green", "black", "red"])
plt.grid()
plt.show
```

```python

# Bubble Plot

plt.scatter(df['Age'], df['Salary'], df['experience']*50, color = 'skyblue', edgecolors="black")
plt.title('Age vs Salary')
plt.xlabel('Age')
plt.ylabel('Salary')

plt.show
```

```python

plt.scatter(df["Age"], df["Salary"], c = df["dept"]. map ({"HR": "yellow", "IT": "blue", "Finance": "orange"}))
plt.xlabel ("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary vs Dept")

plt.show()
```
