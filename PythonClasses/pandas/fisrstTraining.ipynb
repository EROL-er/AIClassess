```python
X = house[['SqFt', 'Bedrooms', 'Bathrooms','GarageCars']]
Y = house['SalePrice']

lr = LinearRegression()

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.20)
lr.fit(x_train, y_train)

yPrep = lr.predict(x_test)
r2 = r2_score(y_test, yPrep)
print(r2)

plt.scatter(y_test, yPrep, color='black')
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.grid()
plt.show()



def predic(SqFt, Bedrooms, Bathrooms,GarageCars):

  return lr.predict([[SqFt, Bedrooms, Bathrooms,GarageCars]])

predic(2730,1 ,1,0)
```
