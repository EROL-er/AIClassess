# Data Group by

```python
groupby_df = df.copy()
category_groups = groupby_df.groupby('category')



category_sum = groupby_df.groupby('category')['total_amount'].sum()
category_sum.sort_values(ascending=False)
```

```python
price_sum = groupby_df.groupby('category')['price'].mean()
price_sum.sort_values(ascending=False).round(2)
groupby_df.groupby('category').size()
```

```python
price_sum = groupby_df.groupby('category')['price'].agg(['count', 'mean','median','std']).round(2)
price_sum
```

```python
price_analysis = groupby_df.groupby ('category', as_index=False).agg(
count=('price', 'count'),
total=('price','sum'),
average=('price', 'mean'),
median=('price', 'median'),
min_price=('price', 'min'),
max_price=('price','max'),
std_dev=('price', 'std')
)
price_analysis.round(2)
```

```python
multi_col_agg = groupby_df.groupby ( 'category',as_index=False) .agg({
'price': ['mean','max','min' ],
'quantity': ['sum','mean' ],
'order_id': 'count'
})
multi_col_agg
```

```python
multi_col_agg = (
groupby_df.groupby('category').agg(
avg_price=('price', 'mean'),
max_price=('price','max'),
min_price=('price','min'),
total_quantity=('quantity', 'sum'),
avg_quantity=('quantity','mean'),
order_count=('order_id','count')
))

multi_col_agg.columns = ['avg_price','max_price', 'min_price', 'total_quantity', 'avg_quantity', 'order_count']
multi_col_agg. reset_index (inplace=True)

multi_col_agg
```

```python
multy_category = groupby_df.groupby(['city','category'])
multy_category['quantity'].sum().unstack(fill_value=0).round(2).reset_index()
```

```python
city_analysis = groupby_df.groupby ('city').agg({
'customer_name': 'nunique',
'order_id': 'count',
'price': 'mean'
}).round(2)

city_analysis.columns = ['Unique_Customers', 'Total_Orders', 'Avg_Order_Value']
city_analysis
top_categories = groupby_df.groupby(['city','price']) ['total_amount']. sum().sort_values(ascending=-False)
top_categories.reset_index().head(10)
```
