```python
merged = pd.merge(pop, abbrevs, how='outer', left_on='state/region', right_on='abbreviation')
merged=merged.drop('abbreviation', axis=1)


merged[merged['state'].isnull()]
merged['state'].isnull().mean()*100
# merged.loc[merged['state'].isnull(),'state/region']
# merged.loc[merged ['state/region'] == 'PR', 'state'] = 'Puerto Rico'
# merged.loc[merged ['state/region'] == 'USA','state']= 'United States'
# # merged.loc[merged ['population'].isna()]= 0

# merged['state'].unique()

merged
```

```python
final = pd.merge(merged, areas, on = 'state', how='left' )
final['state'][final['population'].isnull()]
final.dropna(inplace=True)
final['state'].unique()
```

```python
final[(final['year'] == 2010) & (final['ages']== 'total')]
data2010 = final.query ("year == 2010 & ages == 'total'")
data2010.set_index('state', inplace=True)
data2010=data2010.reset_index()
dansity = data2010['population'] / data2010['area (sq. mi)']
dansity.sort_values(ascending=False)
data2010['density (per sq. mi)'] = data2010['population'] / data2010['area (sq. mi)']
dansity = data2010['population'] / data2010['area (sq. mi)']


```

# Data Visualization

```python
plt.figure(figsize=(6,6))
data2010.head(10).plot(
    x='state',
    y='density (per sq. mi)',
    kind='bar',
    color='skyblue'
)
idx = data2010['density (per sq. mi)'].idxmin()
a=data2010.loc[idx, ['state', 'density (per sq. mi)']]

plt.title(f'{a.loc['state']}')
plt.xlabel('State')
plt.ylabel('Value')
plt.show()

# data2010[data2010['density (per sq. mi)'].max()]
```

```python
null_summary = merged['state'].isnull().value_counts(normalize=True) * 100
null_summary.index = ['Not Null', 'Null']

plt.figure(figsize=(5,5))
null_summary.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    colors=['lightgreen', 'salmon']
)
plt.title('Percentage of Nulls in "state" Column')
plt.ylabel('')
plt.show()
```
