```python
# %pip install scikit-learn
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import scipy.stats as stats
```

```python
plt.style.use("seaborn-v0_8-whitegrid")
sns.set_palette('tab10')
plt.rcParams['figure.figsize'] = [10,6]
plt.rcParams['figure.dpi'] = 100
```

```python
from sklearn.datasets import fetch_california_housing, load_iris, load_diabetes
import ssl

ssl._create_default_https_context = ssl._create_unverified_context

california = fetch_california_housing()
california_df = pd.DataFrame(california.data, columns=california.feature_names)


california_df.head()
```

```python

# --- Pick vectors (columns) ---
v = california_df['MedInc'].values[:10]        # Median income
house_age = california_df['HouseAge'].values[:10]  # House age

# --- Step 1: Compute L2 norm ---
v_norm = np.linalg.norm(v)
print("L2 norm of MedInc:", v_norm)

# --- Step 2: Normalize MedInc ---
v_unit = v / v_norm
print("\nFirst 5 normalized values:", v_unit[:5])

# --- Step 3: Dot product similarity with HouseAge ---
similarity = np.dot(v_unit, house_age / np.linalg.norm(house_age))
print("\nDot product similarity (MedInc vs HouseAge):", similarity)

# --- Step 4: Rank check for multiple features ---
X = california_df[['MedInc', 'HouseAge', 'AveRooms']].values[:50]
rank = np.linalg.matrix_rank(X)
print("\nMatrix Rank:", rank)
```
