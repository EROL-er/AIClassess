```python
df = pd.read_csv('../../Datasets/fromclass/state-population.csv')

def make_df(cols, ind):
  data = {c: [str(c) + str(i+1) for i in ind]
        for c in cols}
  return pd.DataFrame(data, ind)

```

```python
class Display(object):
    def __init__(self, *args):
        self.args = args

    def _repr_html_(self):
        """For Jupyter/Notebook rich HTML display"""
        return ''.join(
            f"<b>In [{a}]</b><br>{eval(a)._repr_html_() if hasattr(eval(a), '_repr_html_') else repr(eval(a))}<br><br>"
            for a in self.args
        )

    def __repr__(self):
        """For standard text/plain display"""
        return '\n\n'.join(
            f"In [{a}]\n{repr(eval(a))}"
            for a in self.args
        )
```
