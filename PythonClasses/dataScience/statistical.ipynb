```python
data = {
    "Name": ["Rob", "Thomas", "Nina", "Mittal", "Sofia", "Mohan", "Tao", "Deepika", "Rafiq", 'Smith'],
    "Height (ft)": [6.2, 5.7, 4.6, 5.4, 5.9, 4.3, 5.1, 5.2, 4.9, 5.0]
}

df = pd.DataFrame(data)

df.describe()

df['zscore'] = (df['Height (ft)'] - df['Height (ft)'].mean()) / df['Height (ft)'].std()



df[(df.zscore <0)  | (df.zscore >1.65)]

dfn = df[(df.zscore >0)]
dfn
```
