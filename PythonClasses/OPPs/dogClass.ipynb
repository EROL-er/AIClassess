```python
class User:
    # 🏗️ Constructor — who the user is
    def __init__(self, name, email, password):
        self.name = name
        self.email = email
        self.password = password
        print(f"✅ User {self.name} created.\n")

    # ⚙️ Methods — what the user can do
    def login(self):
        print(f"{self.name} logged in using {self.email}")

    def change_password(self, new_password):
        self.password = new_password
        print(f"{self.name}'s password changed.\n")


# Ask the user for details
name = input("Enter name: ")
email = input("Enter email: ")
password = input("Enter password: ")

# Create a User object (constructor runs here)
user = User(name, email, password)

# Perform actions with methods
user.login()

new_pass = input("Enter a new password: ")
user.change_password(new_pass)

# Try login again
user.login()
```
