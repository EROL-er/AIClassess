```python
def allValues(**args):
  for key, value in args.items():
    print("name is :", key, "=", value)
    
  # The rest of your function logic follows
  values_list = list(args.values())
  print(values_list[0])

l1 = ['e', 'd', 'z']
l2 = [1, 2, 3]

# Create a dictionary from the two lists using zip()
data = dict(zip(l1, l2))

# Pass the dictionary to the function using the dictionary unpacking operator (**)
allValues(**data)
```

```python
def allValues(**args):
  for key, value in args.items():
    print("name is :", key, "=", value)


l1 = ['e', 'd', 'z']
l2 = [1, 2, 3]

for key, value in zip(l1, l2):
  kwargs = {key: value}
  allValues(**kwargs)
```
