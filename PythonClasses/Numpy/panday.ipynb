```python
dates = np.arange('2025-01-01','2026-01-01', dtype='datetime64[D]')
rainfal = np.random.uniform(0, 20, size=365)

rain_date = np.zeros(365, dtype=[('date', 'datetime64[D]'), ('rainfal', 'f4')])
rain_date['date'] = dates
rain_date['rainfal'] = rainfal

plt.figure(figsize=(10,4))
plt.plot(rain_date['date'], rain_date['rainfal'], label='Rainfall (mm)', color='lightblue')
plt.xlabel("Date")
plt.ylabel("Rainfall (mm)")
plt.title("Daily Rainfall — 2025")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python

dates = np.arange('2025-01-01','2026-01-01', dtype='datetime64[D]')
rainfal = np.random.uniform(0, 20, size=365)

rain_date = np.zeros(365, dtype=[('date', 'datetime64[D]'), ('rainfal', 'f4')])
rain_date['date'] = dates
rain_date['rainfal'] = rainfal

x = np.argmax(rain_date['rainfal'])        
max_date = rain_date['date'][x]              
max_rain = rain_date['rainfal'][x]          

print("Max rainfall:", max_rain, "on", max_date)


```
