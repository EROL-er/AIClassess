```python
data = pd.read_csv('../../Datasets/Lesson_04_Working_With_Pandas/Assisted_Practice_Dataset/Weather_data.csv')
data
Temperature = np.array(data['Temperature'])
day = np.array(data['Day'])

highest_index = Temperature.argmax()
lowest_index = Temperature.argmin()

print('Average Temperature: ', np.round(Temperature.mean())) 
print('Maximum Temperature:  ', np.round(Temperature.max()), "The highest Temperature's day of the month is: ",day[highest_index])
print('Minimum Temperature:  ', np.round(Temperature.min()), "The lowest Temperature's day of the month is: ",day[lowest_index])
```
