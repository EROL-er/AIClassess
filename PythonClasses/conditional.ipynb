```python
#logical Operators
# and, or, not
a=5
b=10
c=15
if a<b and b<c:
    print("a is less than b and b is less than c")
if a>b or b<c:
    print("a is greater than b or b is less than c")
if not a>b:
    print("a is not greater than b")
```
