```python
import csv, ast

with open('../textFiles/dataFile.txt','r') as f:
    List_of_list = f.read()

heading_content = ['Name', 'Age', 'City', 'Email', 'Phone']

with open('../textFiles/writeToFile.csv', 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(heading_content)
    for row in List_of_list:
        w.writerow(row)
```
