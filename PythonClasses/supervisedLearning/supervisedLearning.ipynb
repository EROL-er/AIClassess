# Supervised learning (regression)

```python
# %pip install openpyxl
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

df = pd.read_csv('../../Datasets/fromclass/ecommerce_sales_data.csv')
df3 = pd.read_csv('../../Datasets/fromclass/house.csv')
data = pd.read_excel('../../Datasets/fromclass/head_brain_synthetic_237rows.xlsx')
```

```python
weeks_data = [1, 2, 3, 4, 5]
salas_data = [2, 4, 5, 4, 5]

x=df['price'].head(20)
y=df['total_amount'].head(20)

df2 = pd.DataFrame({'weeks': weeks_data,
                   'salas': salas_data})

X = df2['weeks'].values
Y = df2['salas'].values
mean_x = np.mean(X)
mean_y = np.mean(Y)
n = len(X)
numer = 0 # counter
denom = 0  # counter

for i in range(n):
  numer += (X[i] - mean_x) * (Y[i] - mean_y)
  denom += (X[i] - mean_x) ** 2

m =  numer/denom
c =  mean_y - ( m * mean_x)

print(f'm : {m} ')
print(f'c : {c} ')

max_X = np.max(X) + 1
min_X = np.min(X) - 1
x = np.linspace(min_X, max_X,)
y = (m * x) + c

plt.plot(x, y, color='#8c250f', label='Regression line', linestyle='--')
plt.scatter(X, Y, color="#043078", label='Data point', s=70)
plt.xlabel('Weeks')
plt.ylabel('Sales')
plt.title('LR demo')
plt.grid(True, alpha=0.5, linestyle=":")
plt.legend(loc='best')
np.linspace(min_X, max_X, num=10, endpoint=False)
```

```python
X = X.reshape([n,1])

reg = LinearRegression()
reg = reg.fit(X,Y)
reg
r2 = reg.score(X,Y)
r2
```

```python
              

reg = LinearRegression()
reg.fit(X, Y)

m = reg.coef_[0]
c = reg.intercept_
r2 = reg.score(X, Y)

print("m:", m)
print("c:", c)
print("R²:", r2)

# For plotting the line
x_line = np.linspace(X.min() - 1, X.max()).reshape(-1, 1)
y_line = reg.predict(x_line)

plt.plot(x_line, y_line, '--', label='Regression line')
plt.scatter(X, Y, s=70, label='Data points')
plt.xlabel('Weeks')
plt.ylabel('Sales')
plt.title('LR demo')
plt.grid(True, alpha=0.5, linestyle=':')
plt.legend()
plt.show()
```
