# Introduction: What is Pandas and Why Use It?

## What is Pandas?

Pandas is a powerful, open-source data analysis and manipulation library for Python. 
It provides data structures and functions needed to work with structured data quickly and effectively.

The name "Pandas" comes from "Panel Data" and is also a play on "Python Data Analysis."

## Why Use Pandas for Data Analysis?

1. **Easy Data Loading**: Read data from various formats (CSV, Excel, JSON, SQL databases)
2. **Data Cleaning**: Handle missing values, duplicates, and data type issues
3. **Data Manipulation**: Filter, sort, group, and transform data easily
4. **Data Analysis**: Perform statistical operations and aggregations
5. **Integration**: Works seamlessly with other Python libraries (NumPy, Matplotlib, Scikit-learn)
6. **Performance**: Built on NumPy for fast numerical operations

## Core Data Structures

Pandas provides two main data structures:
- **Series**: 1-dimensional labeled array (like a column in a spreadsheet)
- **DataFrame**: 2-dimensional labeled data structure (like a spreadsheet or SQL table)

## Boolean Indexing (Conditional Selection)

Boolean indexing allows you to filter data based on conditions.

### Creating Boolean Masks Based on Column Values

```python
df.head()
```

```python
expensive_mask = df['price'] > 100
expensive_mask.sum()
```

```python
print("=== Boolean Indexing ===")

# Create boolean masks
print("Boolean mask for expensive products (price > 100):")
expensive_mask = df['price'] > 100
print(expensive_mask.head())
print(f"Type: {type(expensive_mask)}")
print()
```

```python
# Check how many True/False values
print(f"Expensive products: {expensive_mask.sum()}")
print(f"Affordable products: {(~expensive_mask).sum()}")
print()
```


### Applying Masks to Filter DataFrames


```python
print("=== Filtering with Boolean Masks ===")

# Apply the mask to get expensive products
expensive_products = df[expensive_mask]
print("Expensive products (price > 100):")
expensive_products.head()
```

```python
# Direct filtering (more common approach)
affordable_products = df[df['price'] <= 100]
print("Affordable products (price <= 100):")
affordable_products.head()
```

```python
df[df['category'] == 'Books'].head()
```


### Combining Conditions (`& `for `AND`, `|` for `OR`)


```python
print("=== Combining Multiple Conditions ===")

# AND condition (&) - both conditions must be True
# Note: Use parentheses around each condition!
expensive_electronics = df[(df['price'] > 100) & (df['category'] == 'Electronics')]
print("Expensive Electronics (price > 100 AND category = Electronics):")
expensive_electronics.head()
```

```python
# OR condition (|) - either condition can be True
electronics_or_expensive = df[(df['category'] == 'Electronics') | (df['price'] > 500)]
print("Electronics OR expensive products (price > 500):")
electronics_or_expensive.head(5)
```

```python
# Multiple conditions with AND
specific_orders = df[(df['price'] > 50) & (df['quantity'] >= 2) & (df['price'] < 200)]
print("Specific orders (price 50-200 AND quantity >= 2):")
specific_orders.head()
```

```python
# NOT condition (~)

non_electronics = df[~(df['category'] == 'Electronics')]
print("Non-Electronics products:")
non_electronics.head(5)
```

## Handling Missing Data

Missing data is common in real-world datasets. Let's learn how to identify and handle it.

### Identifying Missing Values: `df.isnull().sum()`

```python
print("=== Identifying Missing Data ===")

# Let's add some missing values for demonstration
df_with_missing = df.copy()

# Check for missing values
print("Missing values per column:")
df_with_missing.isnull().sum()
```

```python
# Check percentage of missing values
print("Percentage of missing values:")
(df_with_missing.isnull().sum() / len(df_with_missing)) * 100
```

```python
df_with_missing.isnull().mean()*100
```

```python
# Visualize missing data pattern
print("Rows with any missing values:")
df_with_missing[df_with_missing.isnull().any(axis=1)].head()
```


### Dropping Missing Values: `df.dropna()`


```python
print("=== Dropping Missing Values ===")

# Drop rows with any missing values
df_dropped_rows = df_with_missing.dropna()
print(f"Original shape: {df_with_missing.shape}")
print(f"After dropping rows with missing values: {df_dropped_rows.shape}")
print()
```

```python
# Drop rows only if ALL values are missing
df_dropped_all = df_with_missing.dropna(how='all')
print(f"After dropping rows where ALL values are missing: {df_dropped_all.shape}")
print()
```

```python
# Drop columns with missing values
df_dropped_cols = df_with_missing.dropna(axis=1)
print(f"After dropping columns with missing values: {df_dropped_cols.shape}")
print()
```

```python
# Drop rows with missing values in specific columns
df_dropped_specific = df_with_missing.dropna(subset=['brand'])
print(f"After dropping rows with missing prices: {df_dropped_specific.shape}")
print()
```

```python
# Use thresh parameter - keep rows with at least N non-null values
df_thresh = df_with_missing.dropna(thresh=len(df_with_missing.columns) - 1)  # Allow 1 missing value
print(f"After using thresh (allow 1 missing value): {df_thresh.shape}")
print()
```


### Filling Missing Values: `df.fillna()`


```python
print("=== Filling Missing Values ===")

# Fill with a constant value
df_filled_constant = df_with_missing.fillna('Unknown')
print("After filling with 'Unknown':")
print(df_filled_constant.isnull().sum())
print()
```

```python
# Fill different columns with different values
fill_values = {'customer_name': 'Anonymous Customer', 'price': 0}
df_filled_dict = df_with_missing.fillna(value=fill_values)
print("After filling with dictionary:")
print(df_filled_dict.isnull().sum())
print()
```

```python
# Fill with statistical measures (for numeric columns)
if 'price' in df_with_missing.columns:
    # Fill missing prices with mean
    mean_price = df_with_missing['price'].mean()
    df_filled_mean = df_with_missing.copy()
    df_filled_mean['price'].fillna(mean_price)
    print(f"Missing prices filled with mean ({mean_price:.2f}):")
    print(df_filled_mean['price'].isnull().sum())
    print()
    
    # Fill with median
    median_price = df_with_missing['price'].median()
    df_filled_median = df_with_missing.copy()
    df_filled_median['price'].fillna(median_price)
    print(f"Missing prices filled with median ({median_price:.2f}):")
    print(df_filled_median['price'].isnull().sum())
    print()
```

```python
# Forward fill (use previous value)
df_ffill = df_with_missing.ffill()
print("After forward fill:")
print(df_ffill.isnull().sum())
print()
```

```python
# Check duplicates based on specific columns
if 'order_id' in df_with_dups.columns:
    order_duplicates = df_with_dups.duplicated(subset=['order_id'])
    print(f"Duplicate order IDs: {order_duplicates.sum()}")
    print()
```

__DataFrame__.`duplicated`(*subset=None, keep='first'*)

    Return boolean Series denoting duplicate rows.
    
    Considering certain columns is optional.

__Parameters:__

__subset: *column label or sequence of labels, optional*__
        
    Only consider certain columns for identifying duplicates, by default use all of the columns.

**keep:{*‘first’, ‘last’, False}, default ‘first’***

Determines which duplicates (if any) to mark.
    
* `first` : Mark duplicates as `True` except for the first occurrence.
        
* `last` : Mark duplicates as `True` except for the last occurrence.
    
* `False` : Mark all duplicates as `True`.

```python
# Keep different duplicates (first, last, False)
print("Duplicated (keep='first'):", df_with_dups.duplicated(keep='first').sum())
print("Duplicated (keep='last'):", df_with_dups.duplicated(keep='last').sum())
print("Duplicated (keep=False):", df_with_dups.duplicated(keep=False).sum())
print()
```


### Dropping Duplicates: `df.drop_duplicates()`


```python
print("=== Dropping Duplicates ===")

# Drop duplicate rows
df_no_dups = df_with_dups.drop_duplicates()
print(f"After dropping duplicates: {df_no_dups.shape}")
print()
```

```python
# Drop duplicates based on specific columns
if 'order_id' in df_with_dups.columns:
    df_unique_orders = df_with_dups.drop_duplicates(subset=['order_id'])
    print(f"After dropping duplicate order IDs: {df_unique_orders.shape}")
    print()

# Keep last occurrence instead of first
df_keep_last = df_with_dups.drop_duplicates(keep='last')
print(f"After dropping duplicates (keep='last'): {df_keep_last.shape}")
print()
```

# Module F: Data Grouping & Aggregation

## The `groupby()` Operation

The groupby operation is one of the most powerful features in pandas. It allows you to:
1. **Split** data into groups based on some criteria
2. **Apply** a function to each group independently  
3. **Combine** the results into a data structure

This is known as the **Split-Apply-Combine** paradigm.

### Concept of Split-Apply-Combine

```python

print("=== Understanding Split-Apply-Combine ===")
print("1. SPLIT: Divide data into groups based on some criteria")
print("2. APPLY: Apply a function (like mean, sum, count) to each group")
print("3. COMBINE: Combine the results into a final data structure")
print()

# Prepare data for groupby examples
groupby_df = df.copy()

# Ensure we have the necessary columns
if 'category' not in groupby_df.columns:
    categories = ['Electronics', 'Clothing', 'Books', 'Home & Garden', 'Sports']
    groupby_df['category'] = np.random.choice(categories, len(groupby_df))

if 'city' not in groupby_df.columns:
    cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
    groupby_df['city'] = np.random.choice(cities, len(groupby_df))
```


### Grouping by One Column


```python

print("=== Grouping by Single Column ===")

# Group by category
category_groups = groupby_df.groupby('category')
print(f"GroupBy object: {category_groups}")
print(f"Number of groups: {category_groups.ngroups}")
print("Group keys:", list(category_groups.groups.keys()))
print()

# See the groups
print("Groups and their indices:")
for name, group in category_groups:
    print(f"Group '{name}': {len(group)} rows")
    break  # Just show first group to save space
print()
```


### Applying Aggregation Functions

Let's apply various aggregation functions to our grouped data.


```python

print("=== Aggregation Functions ===")

# Basic aggregations on grouped data
print("1. Sum by category:")
if 'total_amount' in groupby_df.columns:
    category_sum = groupby_df.groupby('category')['total_amount'].sum()
    print(category_sum)
elif 'price' in groupby_df.columns:
    category_sum = groupby_df.groupby('category')['price'].sum()
    print(category_sum)
print()

print("2. Mean by category:")
if 'price' in groupby_df.columns:
    category_mean = groupby_df.groupby('category')['price'].mean()
    print(category_mean.round(2))
print()

print("3. Count by category:")
category_count = groupby_df.groupby('category').size()
print(category_count)
print()

print("4. Multiple statistics:")
if 'price' in groupby_df.columns:
    category_stats = groupby_df.groupby('category')['price'].agg(['count', 'mean', 'median', 'std'])
    print(category_stats.round(2))
print()
```


### Different Aggregation Functions


```python

print("=== Various Aggregation Functions ===")

# sum(): Total values
print("sum() - Total sales by category:")
if 'total_amount' in groupby_df.columns:
    print(groupby_df.groupby('category')['total_amount'].sum().round(2))
print()

# mean(): Average values  
print("mean() - Average price by category:")
if 'price' in groupby_df.columns:
    print(groupby_df.groupby('category')['price'].mean().round(2))
print()

# median(): Middle value
print("median() - Median price by category:")
if 'price' in groupby_df.columns:
    print(groupby_df.groupby('category')['price'].median().round(2))
print()

# count(): Count non-null values
print("count() - Count of orders by category:")
print(groupby_df.groupby('category')['order_id'].count())
print()

# std(): Standard deviation
print("std() - Price standard deviation by category:")
if 'price' in groupby_df.columns:
    print(groupby_df.groupby('category')['price'].std().round(2))
print()

# size(): Count all rows (including null)
print("size() - Total rows by category:")
print(groupby_df.groupby('category').size())
print()
```


### Using agg() for Multiple Operations


```python

print("=== Using agg() for Custom Aggregations ===")

# Apply multiple functions at once
if 'price' in groupby_df.columns:
    price_analysis = groupby_df.groupby('category')['price'].agg({
        'count': 'count',
        'total': 'sum',
        'average': 'mean',
        'median': 'median',
        'min_price': 'min',
        'max_price': 'max',
        'std_dev': 'std'
    })
    print("Comprehensive price analysis by category:")
    print(price_analysis.round(2))
    print()

# Apply different functions to different columns
if all(col in groupby_df.columns for col in ['price', 'quantity']):
    multi_col_agg = groupby_df.groupby('category').agg({
        'price': ['mean', 'max', 'min'],
        'quantity': ['sum', 'mean'],
        'order_id': 'count'
    })
    print("Multiple columns aggregation:")
    print(multi_col_agg.round(2))
    print()
```


### Grouping by Multiple Columns


```python

print("=== Grouping by Multiple Columns ===")

# Group by category and city
multi_group = groupby_df.groupby(['category', 'city'])
print("Grouping by category and city:")
print("Number of groups:", multi_group.ngroups)
print()

# Aggregate multi-group data
if 'price' in groupby_df.columns:
    multi_group_stats = multi_group['price'].mean().round(2)
    print("Average price by category and city:")
    print(multi_group_stats.head(10))  # Show first 10 to save space
    print()

# Unstack to create a pivot-like table
if 'price' in groupby_df.columns:
    pivot_like = multi_group['price'].mean().unstack(fill_value=0).round(2)
    print("Pivot-like table (category vs city):")
    print(pivot_like)
    print()
```


### Practical Examples to Illustrate Power of GroupBy


```python

print("=== Practical Business Examples ===")

# 1. Sales performance by category
print("1. Sales Performance Analysis:")
if all(col in groupby_df.columns for col in ['category', 'total_amount', 'quantity']):
    sales_performance = groupby_df.groupby('category').agg({
        'total_amount': ['sum', 'mean', 'count'],
        'quantity': 'sum'
    }).round(2)
    print(sales_performance)
print()

# 2. Customer analysis by city
print("2. Customer Analysis by City:")
if 'customer_name' in groupby_df.columns:
    city_analysis = groupby_df.groupby('city').agg({
        'customer_name': 'nunique',  # unique customers
        'order_id': 'count',         # total orders
        'price': 'mean'              # average order value
    }).round(2)
    city_analysis.columns = ['Unique_Customers', 'Total_Orders', 'Avg_Order_Value']
    print(city_analysis)
print()

# 3. Product performance analysis
print("3. Top Categories by Revenue:")
if 'total_amount' in groupby_df.columns:
    top_categories = groupby_df.groupby('category')['total_amount'].sum().sort_values(ascending=False)
    print(top_categories.round(2))
print()

# 4. Time-based analysis (if date column exists)
if 'order_date' in groupby_df.columns:
    # Convert to datetime if not already
    groupby_df['order_date'] = pd.to_datetime(groupby_df['order_date'])
    
    # Extract month for analysis
    groupby_df['month'] = groupby_df['order_date'].dt.month
    
    print("4. Monthly Sales Trend:")
    monthly_sales = groupby_df.groupby('month')['total_amount'].sum().round(2)
    print(monthly_sales)
    print()
```
